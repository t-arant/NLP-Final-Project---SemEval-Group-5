```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_1 = pd.read_csv("subtask1_1.csv")
df_1
```

```python
plt.figure(figsize=(8,5))
plt.plot(df_1['Epoch'], df_1['Train Loss'], marker='o', label='Train Loss')
plt.plot(df_1['Epoch'], df_1['Validation Loss'], marker='x', label='Dev Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training and Validation Loss over Epochs')
plt.legend()
plt.grid(True)
plt.show()
```
